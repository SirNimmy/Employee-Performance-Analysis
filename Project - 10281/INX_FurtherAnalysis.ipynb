## WORKINGS:

The provided employee dataset contains 1200 rows and is well-structured for machine learning modeling.
The shape of the data is 1200 × 28, with 16 qualitative features and 11 quantitative features.
Additionally, the Employee ID is not considered a relevant feature for predicting performance ratings.

The dataset is supervised and categorical. The predictor variables are primarily ordinal, with a few being nominal, while the target variable, 'Performance Rating', is ordinal. The steps followed in the project are:

- Imported the dataset, identified the predictor and target variables, and checked for any missing values.

- Analyzed department-wise performance as required.

- Applied label encoding to the ordinal columns.

- Calculated correlation coefficients to study relationships between variables and selected the most important features for analysis.

- Standardized the data and split it into training and testing sets.

- Trained models using Logistic Regression, Decision Tree, Random Forest, and XGBoost Classifier to predict the target variable, and evaluated the models based on their accuracy.

- Exported the model that achieved the highest accuracy for future use.
